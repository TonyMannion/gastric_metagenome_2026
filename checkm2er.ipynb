```python
import glob
import os
import pandas as pd
```

```python
folder_list = glob.glob('/home/dcm/250513Pat_250728Pat_250930Pat/contig_bins/*')

for folder in folder_list:
    filename = folder.split("/")[-1]
    print("\n")
    print(filename)
    print("\n")
    os.system(f'/home/dcm/checkm2/bin/checkm2 predict \
    --input {folder} \
    -o /home/dcm/250513Pat_250728Pat_250930Pat/checkm2_contig_bins/{filename} \
    --threads 90 --force -x .fasta --remove_intermediates')
```

```python

file_list = glob.glob('/home/dcm/250513Pat_250728Pat_250930Pat/checkm2_contig_bins/*/*tsv')

final_df = pd.DataFrame()

for file in file_list:
    df = pd.read_csv(f'{file}', sep = '\t')
    file_name = file.split('/')[-2]
    df['File'] = file_name   
    df['Species'] = df['Name'].str.split('_').str[2:].str.join('_')
    final_df = pd.concat([final_df, df], ignore_index=True)
    
final_df.to_csv("/home/dcm/250513Pat_250728Pat_250930Pat/checkm2_contig_bins/checkm2_concat.txt", sep='\t', index = False)
```
