```python
###genus level
```

```python
#P vs N - 20 + 26 years
df_mgx = pd.read_csv('stats_out/fisher_mgx_genus_N_vs_P_results.txt', sep = '\t')
df_culture =  pd.read_csv('stats_out/fisher_culture_genus_N_vs_P_results.txt', sep = '\t')
df_16s = pd.read_csv('stats_out/ttest_16S_N_vs_P_results.txt', sep = '\t')

dfm = pd.merge(df_mgx, df_culture, on = 'otu', how = 'outer', suffixes=('_mgx', '_culture'))

dfm = pd.merge(dfm, df_16s, left_on = 'otu', right_on = 'genus', how = 'outer', suffixes=('', '_16S'))

filter_value = 0.05

dfm = dfm[
    (dfm['P-value_mgx'] <= filter_value) &
    (dfm['P-value_culture'] <= filter_value) &
    (dfm['P-value'] <= filter_value)
]

dfm
```
